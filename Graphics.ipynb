```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
def read(micro, nume):
    O_xx = []
    O_yy = []
    O_zz = []
    Y_xx = []
    Y_yy = []
    Y_zz = []
    P_xx = []
    P_yy = []
    P_zz = []
    for i, num in enumerate(nume):
        s11 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_xx/sigma.csv", header=None).iloc[-1, 0])
        O_xx.append(s11/0.1)
        E_x = s11/0.1
            
        s22 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_yy/sigma.csv", header=None).iloc[-1, 1])
        O_yy.append(s22/0.1)
        E_y = s22/0.1
            
        s33 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_zz/sigma.csv", header=None).iloc[-1, 2])
        O_zz.append(s33/0.1)
        E_z = s33/0.1

        # loading zz
        sxx_z = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_zz/sigma_f.csv", header=None).iloc[-1, 0])
        syy_z = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_zz/sigma_f.csv", header=None).iloc[-1, 1])
        sxN_z = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_zz/sigma_f.csv", header=None).iloc[-1, 3])
        syN_z = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_zz/sigma_f.csv", header=None).iloc[-1, 4])

        K0_z = ((sxx_z + syy_z + sxN_z + syN_z)/4)/(s33)
        P_vz = K0_z / (1 + K0_z)
        P_zz.append(P_vz)
    
        Y_z = E_z * (1 + P_vz) * (1 - 2 * P_vz)/ (1- P_vz)
        Y_zz.append(Y_z)

        # loading yy
        sxx_y = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_yy/sigma_f.csv", header=None).iloc[-1, 0])
        szz_y = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_yy/sigma_f.csv", header=None).iloc[-1, 2])
        sxN_y = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_yy/sigma_f.csv", header=None).iloc[-1, 3])
        szN_y = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_yy/sigma_f.csv", header=None).iloc[-1, 5])

        K0_y = ((sxx_y + szz_y + sxN_y + szN_y)/4)/(s22)
        P_vy = K0_y / (1 + K0_y)
        P_yy.append(P_vy)
    
        Y_y = E_y * (1 + P_vy) * (1 - 2 * P_vy)/ (1- P_vy)
        Y_yy.append(Y_y)


        # loading xx
        syy_x = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_xx/sigma_f.csv", header=None).iloc[-1, 1])
        szz_x = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_xx/sigma_f.csv", header=None).iloc[-1, 2])
        syN_x = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_xx/sigma_f.csv", header=None).iloc[-1, 4])
        szN_x = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/oedometric_xx/sigma_f.csv", header=None).iloc[-1, 5])

        K0_x = ((syy_x + szz_x + syN_x + szN_x)/4)/(s11)
        P_vx = K0_x / (1 + K0_x)
        P_xx.append(P_vx)
    
        Y_x = E_x * (1 + P_vx) * (1 - 2 * P_vx)/ (1- P_vx)
        Y_xx.append(Y_x)

            
    return O_xx, Y_xx, P_xx, O_yy, Y_yy, P_yy, O_zz, Y_zz, P_zz


def read_iso(micro, nume):
    K = []
    for i, num in enumerate(nume):
        s11 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/isotropic/sigma.csv", header=None).iloc[-1, 0])
        s22 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/isotropic/sigma.csv", header=None).iloc[-1, 1])
        s33 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/mech/isotropic/sigma.csv", header=None).iloc[-1, 2])
        K_ = ((s11 + s22 + s33)/3)/0.1
        K.append(K_)
    return K


def read_k(micro, nume):
    K_x = []
    K_y = []
    K_z = []
    for i, num in enumerate(nume):
        k1 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/hyd/xx/flux.csv", header=None).iloc[-1, 0])
        k2 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/hyd/yy/flux.csv", header=None).iloc[-1, 1])
        k3 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/hyd/zz/flux.csv", header=None).iloc[-1, 2])
        K_1 = -0.001/1000*k1 
        K_2 = -0.001/1000*k2
        K_3 = -0.001/1000*k3
        K_x.append(K_1)
        K_y.append(K_2)
        K_z.append(K_3)
    return K_x, K_y, K_z

def read_l(micro, nume):
    l_x = []
    l_y = []
    l_z = []
    for i, num in enumerate(nume):
        l1 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/the/xx/flux.csv", header=None).iloc[-1, 0])
        l2 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/the/yy/flux.csv", header=None).iloc[-1, 1])
        l3 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/the/zz/flux.csv", header=None).iloc[-1, 2])
        l_1 = -l1 
        l_2 = -l2
        l_3 = -l3
        l_x.append(l_1)
        l_y.append(l_2)
        l_z.append(l_3)
    return l_x, l_y, l_z

def read_biot(micro, nume):
    biot = []
    for i, num in enumerate(nume):
        s11 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressP/sigma.csv", header=None).iloc[-1, 0])
        s22 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressP/sigma.csv", header=None).iloc[-1, 1])
        s33 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressP/sigma.csv", header=None).iloc[-1, 2])
        Biot_ = -((s11 + s22 + s33)/3)/0.1
        biot.append(Biot_)
    return biot

def read_C(micro, nume):
    C = []
    for i, num in enumerate(nume):
        s11 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressT/sigma.csv", header=None).iloc[-1, 0])
        s22 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressT/sigma.csv", header=None).iloc[-1, 1])
        s33 = (pd.read_csv(f"SIMULATIONS/{micro}/dict_fft_{num}/pressT/sigma.csv", header=None).iloc[-1, 2])
        C_ = -((s11 + s22 + s33)/3)/1
        C.append(C_)
    return C
```
